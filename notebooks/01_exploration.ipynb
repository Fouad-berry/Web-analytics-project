```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import load_sessions, load_events, load_transactions
from src.utils import set_style, print_section

set_style()
sessions = load_sessions()
events = load_events()
trans = load_transactions()

print(f'Sessions     : {len(sessions):,}')
print(f'Events       : {len(events):,}')
print(f'Transactions : {len(trans):,}')
```

## KPIs rapides

```python
print_section('KPIs globaux')
print(f"Période : {sessions['session_date'].min().date()} → {sessions['session_date'].max().date()}")
print(f"Users uniques : {sessions['user_id'].nunique():,}")
print(f"Conversion rate : {100*sessions['converted'].mean():.2f}%")
print(f"Bounce rate : {100*sessions['bounced'].mean():.2f}%")
print(f"Revenue total : ${trans['revenue_usd'].sum():,.0f}")
print(f"AOV : ${trans['revenue_usd'].sum()/sessions['converted'].sum():,.2f}")
```

## Répartition par canal

```python
sessions['channel'].value_counts().plot(kind='barh', figsize=(10, 5),
                                         color='steelblue', edgecolor='black')
plt.title('Sessions par canal d\'acquisition')
plt.xlabel('Sessions'); plt.show()
```
